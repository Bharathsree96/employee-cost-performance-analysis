```python
import pandas as pd
import numpy as np
```

```python
# Task 1 - creating dataframes and saving as csv

project_df = pd.DataFrame({
    'ID': ['A001','A002','A003','A004','A005','A002','A005','A003','A001','A003','A001','A004','A004','A005'],
    'Project': ['Project 1','Project 2','Project 3','Project 4','Project 5','Project 6','Project 7','Project 8','Project 9','Project 10','Project 11','Project 12','Project 13','Project 14'],
    'Cost': [1002000,2000000,4500000,5500000,np.nan,680000,400000,350000,np.nan,300000,2000000,1000000,3000000,200000],
    'Status': ['Finished','Ongoing','Finished','Ongoing','Finished','Failed','Finished','Failed','Ongoing','Finished','Failed','Ongoing','Finished','Finished']
})

employee_df = pd.DataFrame({
    'ID': ['A001','A002','A003','A004','A005'],
    'Name': ['John Alter','Alice Luxumberg','Tom Sabestine','Nina Adgra','Amy Johny'],
    'Gender': ['M','F','M','F','F'],
    'City': ['Paris','London','Berlin','Newyork','Madrid'],
    'Age': [25,27,29,31,30]
})

seniority_df = pd.DataFrame({
    'ID': ['A001','A002','A003','A004','A005'],
    'Designation Level': [2,2,3,2,3]
})

project_df.to_csv('project.csv', index=False)
employee_df.to_csv('employee.csv', index=False)
seniority_df.to_csv('seniority.csv', index=False)

print("Task 1 Done")
print(project_df)
print(employee_df)
print(seniority_df)
```

```python
# Task 2 - fill missing cost using running average with for loop

project_df = pd.read_csv('project.csv')
employee_df = pd.read_csv('employee.csv')
seniority_df = pd.read_csv('seniority.csv')

total = 0
count = 0
for i in range(len(project_df)):
    if pd.isna(project_df.loc[i,'Cost']):
        project_df.loc[i,'Cost'] = total / count
    else:
        total = total + project_df.loc[i,'Cost']
        count = count + 1

print("Task 2 Done")
print(project_df)
```

```python
# Task 3 - split name into first name and last name

employee_df['First Name'] = employee_df['Name'].str.split(' ').str[0]
employee_df['Last Name'] = employee_df['Name'].str.split(' ').str[1]
employee_df = employee_df.drop('Name', axis=1)

print("Task 3 Done")
print(employee_df)
```

```python
# Task 4 - merge all dataframes into final

final = pd.merge(project_df, employee_df, on='ID')
final = pd.merge(final, seniority_df, on='ID')

print("Task 4 Done")
print(final)
```

```python
# Task 5 - add 5% bonus for finished projects

final['Bonus'] = 0
for i in final.index:
    if final.loc[i,'Status'] == 'Finished':
        final.loc[i,'Bonus'] = final.loc[i,'Cost'] * 0.05

print("Task 5 Done")
print(final)
```

```python
# Task 6 - demote failed projects and remove designation above 4

for i in final.index:
    if final.loc[i,'Status'] == 'Failed':
        final.loc[i,'Designation Level'] = final.loc[i,'Designation Level'] + 1

final = final[final['Designation Level'] <= 4]

print("Task 6 Done")
print(final)
```

```python
# Task 7 - add Mr. Mrs. and drop gender column

for i in final.index:
    if final.loc[i,'Gender'] == 'M':
        final.loc[i,'First Name'] = 'Mr. ' + final.loc[i,'First Name']
    else:
        final.loc[i,'First Name'] = 'Mrs. ' + final.loc[i,'First Name']

final = final.drop('Gender', axis=1)

print("Task 7 Done")
print(final)
```

```python
# Task 8 - promote designation if age more than 29

for i in final.index:
    if final.loc[i,'Age'] > 29:
        final.loc[i,'Designation Level'] = final.loc[i,'Designation Level'] - 1

print("Task 8 Done")
print(final)
```

```python
# Task 9 - total project cost per employee

TotalProjCost = final.groupby(['ID','First Name'])['Cost'].sum().reset_index()
TotalProjCost.columns = ['ID','First Name','Total Cost']

print("Task 9 Done")
print(TotalProjCost)
```

```python
# Task 10 - employees whose city has letter o

result = final[final['City'].str.contains('o')]

print("Task 10 Done")
print(result)
```
